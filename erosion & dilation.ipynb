```python
#Import required Modules
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def load_img():
    blank_img = np.zeros((600, 850), dtype=np.uint8)
    font = cv2.FONT_HERSHEY_SCRIPT_COMPLEX
    cv2.putText(blank_img,text='LOHINI',org=(175,359),fontFace=font,fontScale= 5,color=(255,255,255),thickness=25,lineType=cv2.LINE_AA)
    return blank_img
```

```python
def display_img(img):
    fig = plt.figure(figsize=(12, 10))
    ax = fig.add_subplot(111)
    ax.imshow(img, cmap="gray")
    plt.show()
```

```python
img = load_img()
```

```python
display_img(img)
```

```python
#Structuring element
img = load_img()
kernel = np.ones((5,5),np.uint8)
erosion5 = cv2.erode( img,kernel,iterations = 4)
```

```python
#Erosion
plt.figure(figsize=(10, 6))
plt.imshow(cv2.cvtColor(erosion5, cv2.COLOR_BGR2RGB))
plt.title("Eroded Image")
plt.show()
```

```python
#Apply the image dilation
dilation5 = cv2.dilate(img, kernel, iterations=4)

# Display the dilated image
plt.figure(figsize=(10, 6))
plt.imshow(cv2.cvtColor(dilation5, cv2.COLOR_BGR2RGB))
plt.title("Dilated Image")
plt.show()
```

```python
plt.figure(figsize=(15, 5))

# Original Image
plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")

# Erosion
plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(erosion5, cv2.COLOR_BGR2RGB))
plt.title("Erosion")

# Dilation
plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(dilation5, cv2.COLOR_BGR2RGB))
plt.title("Dilation")

plt.tight_layout()
plt.show()
```
